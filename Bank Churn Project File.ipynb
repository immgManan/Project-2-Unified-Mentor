Clean Dataset

```python
# remove unnecessary columns
base_data.drop(columns = ['CustomerId', 'Surname'], inplace = True)
```

```python
# Encode Categorical Variables
# for Gender Label encoding is used as it has only two categories
from sklearn.preprocessing import LabelEncoder
base_data['Gender'] = LabelEncoder().fit_transform(base_data['Gender'])

# for Geography One hot encoding is used, more than two categories are present
base_data = pd.get_dummies(base_data, columns = ['Geography'], drop_first = True)
```

```python
base_data.head() # France can be predicted by the absence of Germany and Spain # Female is 0 and Male is 1.
```

```python
# FEATURE ENGINEERING
base_data['Balance_to_Salary_ratio'] = base_data['Balance'] / base_data['EstimatedSalary']

base_data['Product_Density_Indicator'] = base_data['NumOfProducts']/ base_data['Age']

base_data['Engagement_product_indicator'] = base_data['NumOfProducts']*2 + base_data['HasCrCard'] + base_data['IsActiveMember']

base_data['Age_tenure_interaction'] = base_data['Age']*base_data['Tenure']
```

```python
base_data.head()
```

```python
# Scale Numerical Features
from sklearn.preprocessing  import StandardScaler

cols = ['CreditScore', 'Age', 'Balance', 'EstimatedSalary', 'Balance_to_Salary_ratio']

scalar = StandardScaler()

base_data[cols] = scalar.fit_transform(base_data[cols])
```
